```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python

aasist_zh_result='change this to your AASIST zh_eval_score.txt'
aasist_en_result='change this to your AASIST en_eval_score.txt'
aasist_column_name=['audio_file','source','label','score']
aasist_zh_df=pd.read_csv(aasist_zh_result,delimiter=' ',names=aasist_column_name)
aasist_en_df=pd.read_csv(aasist_en_result,delimiter=' ',names=aasist_column_name)

aasist_zh_labels=aasist_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
aasist_zh_scores=aasist_zh_df['score']
aasist_en_labels=aasist_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
aasist_en_scores=aasist_en_df['score']

aasist_fpr_zh, aasist_tpr_zh, aasist_thresholds_zh = roc_curve(aasist_zh_labels, aasist_zh_scores)
aasist_fpr_en, aasist_tpr_en, aasist_thresholds_en = roc_curve(aasist_en_labels, aasist_en_scores)

aasist_eer_zh, aasist_eer_threshold_zh = compute_eer(aasist_fpr_zh, aasist_tpr_zh, aasist_thresholds_zh)
aasist_eer_en, aasist_eer_threshold_en = compute_eer(aasist_fpr_en, aasist_tpr_en, aasist_thresholds_en)

aasist_auc_zh = auc(aasist_fpr_zh, aasist_tpr_zh)
aasist_auc_en = auc(aasist_fpr_en, aasist_tpr_en)


print(f'AASIST zh EER: {aasist_eer_zh}, AUC: {aasist_auc_zh}')
print(f'AASIST en EER: {aasist_eer_en}, AUC: {aasist_auc_en}')
```
