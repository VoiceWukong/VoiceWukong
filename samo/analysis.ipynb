```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python

samo_zh_result='/home/ydoit/AIGC/Eva/samo/samo/samo/zh_eval_score.txt'
samo_en_result='/home/ydoit/AIGC/Eva/samo/samo/samo/en_eval_score.txt'
samo_column_name=['audio_file','flag','label','score','source']
samo_zh_df=pd.read_csv(samo_zh_result,delimiter=' ',names=samo_column_name)
samo_en_df=pd.read_csv(samo_en_result,delimiter=' ',names=samo_column_name)
samo_zh_labels=samo_zh_df['label'].apply(lambda x: 1 if x == 0 else 0)
samo_zh_scores=samo_zh_df['score']
samo_en_labels=samo_en_df['label'].apply(lambda x: 1 if x == 0 else 0)
samo_en_scores=samo_en_df['score']
samo_fpr_zh, samo_tpr_zh, samo_thresholds_zh = roc_curve(samo_zh_labels, samo_zh_scores)
samo_fpr_en, samo_tpr_en, samo_thresholds_en = roc_curve(samo_en_labels, samo_en_scores)
samo_eer_zh, samo_eer_threshold_zh = compute_eer(samo_fpr_zh, samo_tpr_zh, samo_thresholds_zh)
samo_eer_en, samo_eer_threshold_en = compute_eer(samo_fpr_en, samo_tpr_en, samo_thresholds_en)
samo_auc_zh = auc(samo_fpr_zh, samo_tpr_zh)
samo_auc_en = auc(samo_fpr_en, samo_tpr_en)

print(f'SAMO zh EER: {samo_eer_zh}, AUC: {samo_auc_zh}')
print(f'SAMO en EER: {samo_eer_en}, AUC: {samo_auc_en}')
```
