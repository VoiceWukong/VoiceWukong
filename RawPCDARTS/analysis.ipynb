```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
rawpc_zh_result='/home/ydoit/AIGC/Eva/pc-darts/pc-darts-anti-spoofing/zh_eval_score.txt'
rawpc_en_result='/home/ydoit/AIGC/Eva/pc-darts/pc-darts-anti-spoofing/en_eval_score.txt'
rawpc_column_name=['variant','audio_file','flag','source','label','score']
rawpc_zh_df=pd.read_csv(rawpc_zh_result,delimiter=' ',names=rawpc_column_name)
rawpc_en_df=pd.read_csv(rawpc_en_result,delimiter=' ',names=rawpc_column_name)
rawpc_zh_labels=rawpc_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
rawpc_zh_scores=rawpc_zh_df['score']
rawpc_en_labels=rawpc_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
rawpc_en_scores=rawpc_en_df['score']
rawpc_fpr_zh, rawpc_tpr_zh, rawpc_thresholds_zh = roc_curve(rawpc_zh_labels, rawpc_zh_scores)
rawpc_fpr_en, rawpc_tpr_en, rawpc_thresholds_en = roc_curve(rawpc_en_labels, rawpc_en_scores)
rawpc_eer_zh, rawpc_eer_threshold_zh = compute_eer(rawpc_fpr_zh, rawpc_tpr_zh, rawpc_thresholds_zh)
rawpc_eer_en, rawpc_eer_threshold_en = compute_eer(rawpc_fpr_en, rawpc_tpr_en, rawpc_thresholds_en)
rawpc_auc_zh = auc(rawpc_fpr_zh, rawpc_tpr_zh)
rawpc_auc_en = auc(rawpc_fpr_en, rawpc_tpr_en)

print(f'Raw PC-DARTS zh EER: {rawpc_eer_zh}, AUC: {rawpc_auc_zh}')
print(f'Raw PC-DARTS en EER: {rawpc_eer_en}, AUC: {rawpc_auc_en}')
```
