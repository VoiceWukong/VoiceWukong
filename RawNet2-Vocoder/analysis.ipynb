```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
vocoder_zh_result='change this to your RawNet2-Vocoder zh_eval_score.txt path'
vocoder_en_result='change this to your RawNet2-Vocoder eval_score.txt path'
vocoder_column_name=['score1','score2','variant','audio_file','flag','source','label']
vocoder_zh_df=pd.read_csv(vocoder_zh_result,delimiter=' ',names=vocoder_column_name)
vocoder_en_df=pd.read_csv(vocoder_en_result,delimiter=' ',names=vocoder_column_name)
vocoder_zh_labels=vocoder_zh_df['label'].apply(lambda x: 1 if x == 'bonafida' else 0)
vocoder_zh_scores=vocoder_zh_df['score2']
vocoder_en_labels=vocoder_en_df['label'].apply(lambda x: 1 if x == 'bonafida' else 0)
vocoder_en_scores=vocoder_en_df['score2']
vocoder_fpr_zh, vocoder_tpr_zh, vocoder_thresholds_zh = roc_curve(vocoder_zh_labels, vocoder_zh_scores)
vocoder_fpr_en, vocoder_tpr_en, vocoder_thresholds_en = roc_curve(vocoder_en_labels, vocoder_en_scores)
vocoder_eer_zh, vocoder_eer_threshold_zh = compute_eer(vocoder_fpr_zh, vocoder_tpr_zh, vocoder_thresholds_zh)
vocoder_eer_en, vocoder_eer_threshold_en = compute_eer(vocoder_fpr_en, vocoder_tpr_en, vocoder_thresholds_en)
vocoder_auc_zh = auc(vocoder_fpr_zh, vocoder_tpr_zh)
vocoder_auc_en = auc(vocoder_fpr_en, vocoder_tpr_en)
print(f'Vocoder zh EER: {vocoder_eer_zh}, AUC: {vocoder_auc_zh}')
print(f'Vocoder en EER: {vocoder_eer_en}, AUC: {vocoder_auc_en}')
```
