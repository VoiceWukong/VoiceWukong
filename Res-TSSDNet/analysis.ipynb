```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
resnet_zh_result='change this to your Res-TSSDNet zh_eval_score.txt'
resnet_en_result='change this to your Res-TSSDNet en_eval_score.txt'
resnet_column_name=['score1','score2','label','audio_file','source','variant']
resnet_zh_df=pd.read_csv(resnet_zh_result,delimiter=' ',names=resnet_column_name)
resnet_en_df=pd.read_csv(resnet_en_result,delimiter=' ',names=resnet_column_name)
resnet_zh_labels=resnet_zh_df['label']#.apply(lambda x: 1 if int(x) == 0 else 0)
resnet_zh_scores=resnet_zh_df['score1']
resnet_en_labels=resnet_en_df['label']#.apply(lambda x: 1 if int(x) == 0 else 0)
resnet_en_scores=resnet_en_df['score1']
resnet_fpr_zh, resnet_tpr_zh, resnet_thresholds_zh = roc_curve(resnet_zh_labels, resnet_zh_scores)
resnet_fpr_en, resnet_tpr_en, resnet_thresholds_en = roc_curve(resnet_en_labels, resnet_en_scores)
resnet_eer_zh, resnet_eer_threshold_zh = compute_eer(resnet_fpr_zh, resnet_tpr_zh, resnet_thresholds_zh)
resnet_eer_en, resnet_eer_threshold_en = compute_eer(resnet_fpr_en, resnet_tpr_en, resnet_thresholds_en)
resnet_auc_zh = auc(resnet_fpr_zh, resnet_tpr_zh)
resnet_auc_en = auc(resnet_fpr_en, resnet_tpr_en)
print(f'ResNet zh EER: {resnet_eer_zh}, AUC: {resnet_auc_zh}')
print(f'ResNet en EER: {resnet_eer_en}, AUC: {resnet_auc_en}')
```
