```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python

aasist2_zh_result='/home/ydoit/AIGC/Eva/aasist2/SSL_Anti-spoofing/zh_eval_score.txt'
aasist2_en_result='/home/ydoit/AIGC/Eva/aasist2/SSL_Anti-spoofing/en_eval_score.txt'
aasist2_column_name=['variant','audio_file','flag','source','label','score']
aasist2_zh_df=pd.read_csv(aasist2_zh_result,delimiter=' ',names=aasist2_column_name)
aasist2_en_df=pd.read_csv(aasist2_en_result,delimiter=' ',names=aasist2_column_name)
aasist2_zh_labels=aasist2_zh_df['label']
aasist2_zh_scores=aasist2_zh_df['score']
aasist2_en_labels=aasist2_en_df['label']
aasist2_en_scores=aasist2_en_df['score']
aasist2_fpr_zh, aasist2_tpr_zh, aasist2_thresholds_zh = roc_curve(aasist2_zh_labels, aasist2_zh_scores)
aasist2_fpr_en, aasist2_tpr_en, aasist2_thresholds_en = roc_curve(aasist2_en_labels, aasist2_en_scores)
aasist2_eer_zh, aasist2_eer_threshold_zh = compute_eer(aasist2_fpr_zh, aasist2_tpr_zh, aasist2_thresholds_zh)
aasist2_eer_en, aasist2_eer_threshold_en = compute_eer(aasist2_fpr_en, aasist2_tpr_en, aasist2_thresholds_en)
aasist2_auc_zh = auc(aasist2_fpr_zh, aasist2_tpr_zh)
aasist2_auc_en = auc(aasist2_fpr_en, aasist2_tpr_en)


print(f'AASIST2 zh EER: {aasist2_eer_zh}, AUC: {aasist2_auc_zh}')
print(f'AASIST2 en EER: {aasist2_eer_en}, AUC: {aasist2_auc_en}')
```
