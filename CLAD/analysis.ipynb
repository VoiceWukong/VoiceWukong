```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
clad_zh_result='change this to your CLAD zh_eval_score.txt' 
clad_en_result='change this to your CLAD en_eval_score.txt'
clad_column_name=['variant','audio_file','flag','source','label','score']
clad_zh_df=pd.read_csv(clad_zh_result,delimiter=' ',names=clad_column_name)
clad_en_df=pd.read_csv(clad_en_result,delimiter=' ',names=clad_column_name)
clad_zh_labels=clad_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
clad_zh_scores=clad_zh_df['score']
clad_en_labels=clad_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
clad_en_scores=clad_en_df['score']
clad_fpr_zh, clad_tpr_zh, clad_thresholds_zh = roc_curve(clad_zh_labels, clad_zh_scores)
clad_fpr_en, clad_tpr_en, clad_thresholds_en = roc_curve(clad_en_labels, clad_en_scores)
clad_eer_zh, clad_eer_threshold_zh = compute_eer(clad_fpr_zh, clad_tpr_zh, clad_thresholds_zh)
clad_eer_en, clad_eer_threshold_en = compute_eer(clad_fpr_en, clad_tpr_en, clad_thresholds_en)
clad_auc_zh = auc(clad_fpr_zh, clad_tpr_zh)
clad_auc_en = auc(clad_fpr_en, clad_tpr_en)
print(f'CLAD zh EER: {clad_eer_zh}, AUC: {clad_auc_zh}')
print(f'CLAD en EER: {clad_eer_en}, AUC: {clad_auc_en}')
```
