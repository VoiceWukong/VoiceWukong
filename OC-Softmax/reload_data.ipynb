```python
import pickle
from librosa.util import find_files
import scipy.io as sio
import os
```

```python
path='/home/ydoit/AIGC/Eva/OC-softmaxq/AIR-ASVspoof/ZH_Features/eval'
for root,dir,files in os.walk(path):
    for file in files:
        if file.endswith('.mat'):
            file_path=os.path.join(root,file)
            lfcc=sio.loadmat(file_path,verify_compressed_data_integrity=False)['x']
            with open(file_path.replace('.mat','.pkl'),'wb') as f:
                pickle.dump(lfcc,f, protocol=pickle.HIGHEST_PROTOCOL)
            
```
