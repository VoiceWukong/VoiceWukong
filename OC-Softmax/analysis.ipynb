```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
oc_zh_result='change this to path to OC-Softmax zh_eval_score.txt'
oc_en_result='change this to path to OC-Softmax en_eval_score.txt'
oc_column_name=['variant','audio_file','flag','source','label','score']
oc_zh_df=pd.read_csv(oc_zh_result,delimiter=' ',names=oc_column_name)
oc_en_df=pd.read_csv(oc_en_result,delimiter=' ',names=oc_column_name)
oc_zh_labels=oc_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
oc_zh_scores=oc_zh_df['score']
oc_en_labels=oc_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
oc_en_scores=oc_en_df['score']
oc_fpr_zh, oc_tpr_zh, oc_thresholds_zh = roc_curve(oc_zh_labels, oc_zh_scores)
oc_fpr_en, oc_tpr_en, oc_thresholds_en = roc_curve(oc_en_labels, oc_en_scores)
oc_eer_zh, oc_eer_threshold_zh = compute_eer(oc_fpr_zh, oc_tpr_zh, oc_thresholds_zh)
oc_eer_en, oc_eer_threshold_en = compute_eer(oc_fpr_en, oc_tpr_en, oc_thresholds_en)
oc_auc_zh = auc(oc_fpr_zh, oc_tpr_zh)
oc_auc_en = auc(oc_fpr_en, oc_tpr_en)


print(f'OC-Softmax zh EER: {oc_eer_zh}, AUC: {oc_auc_zh}')
print(f'OC-Softmax EER: {oc_eer_en}, AUC: {oc_auc_en}')
```
