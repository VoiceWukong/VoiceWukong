```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
rawnet2_zh_result='/home/ydoit/AIGC/Eva/rawnet2/rawnet2-antispoofing/zh_eval_scores.txt'
rawnet2_en_result='/home/ydoit/AIGC/Eva/rawnet2/rawnet2-antispoofing/en_eval_scores.txt'
rawnet2_column_name=['audio_file','source','label','score']
rawnet2_zh_df=pd.read_csv(rawnet2_zh_result,delimiter=' ',names=rawnet2_column_name)
rawnet2_en_df=pd.read_csv(rawnet2_en_result,delimiter=' ',names=rawnet2_column_name)
rawnet2_zh_labels=rawnet2_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
rawnet2_zh_scores=rawnet2_zh_df['score']
rawnet2_en_labels=rawnet2_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
rawnet2_en_scores=rawnet2_en_df['score']
rawnet2_fpr_zh, rawnet2_tpr_zh, rawnet2_thresholds_zh = roc_curve(rawnet2_zh_labels, rawnet2_zh_scores)
rawnet2_fpr_en, rawnet2_tpr_en, rawnet2_thresholds_en = roc_curve(rawnet2_en_labels, rawnet2_en_scores)
rawnet2_eer_zh, rawnet2_eer_threshold_zh = compute_eer(rawnet2_fpr_zh, rawnet2_tpr_zh, rawnet2_thresholds_zh)
rawnet2_eer_en, rawnet2_eer_threshold_en = compute_eer(rawnet2_fpr_en, rawnet2_tpr_en, rawnet2_thresholds_en)
rawnet2_auc_zh = auc(rawnet2_fpr_zh, rawnet2_tpr_zh)
rawnet2_auc_en = auc(rawnet2_fpr_en, rawnet2_tpr_en)
print(f'RawNet2 zh EER: {rawnet2_eer_zh}, AUC: {rawnet2_auc_zh}')
print(f'RawNet2 en EER: {rawnet2_eer_en}, AUC: {rawnet2_auc_en}')
```
