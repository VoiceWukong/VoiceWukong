```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
rawboost_zh_result='change this to your RawBoost zh_eval_score.txt path'
rawboost_en_result='change this to your RawBoost en_eval_score.txt path'
rawboost_column_name=['variant','audio_file','flag','source','score','label']
rawboost_zh_df=pd.read_csv(rawboost_zh_result,delimiter=' ',names=rawboost_column_name)
rawboost_en_df=pd.read_csv(rawboost_en_result,delimiter=' ',names=rawboost_column_name)
rawboost_zh_labels=rawboost_zh_df['label']
rawboost_zh_scores=rawboost_zh_df['score']
rawboost_en_labels=rawboost_en_df['label']
rawboost_en_scores=rawboost_en_df['score']
rawboost_fpr_zh, rawboost_tpr_zh, rawboost_thresholds_zh = roc_curve(rawboost_zh_labels, rawboost_zh_scores)
rawboost_fpr_en, rawboost_tpr_en, rawboost_thresholds_en = roc_curve(rawboost_en_labels, rawboost_en_scores)
rawboost_eer_zh, rawboost_eer_threshold_zh = compute_eer(rawboost_fpr_zh, rawboost_tpr_zh, rawboost_thresholds_zh)
rawboost_eer_en, rawboost_eer_threshold_en = compute_eer(rawboost_fpr_en, rawboost_tpr_en, rawboost_thresholds_en)
rawboost_auc_zh = auc(rawboost_fpr_zh, rawboost_tpr_zh)
rawboost_auc_en = auc(rawboost_fpr_en, rawboost_tpr_en)
print(f'RawBoost zh EER: {rawboost_eer_zh}, AUC: {rawboost_auc_zh}')
print(f'RawBoost en EER: {rawboost_eer_en}, AUC: {rawboost_auc_en}')
```
