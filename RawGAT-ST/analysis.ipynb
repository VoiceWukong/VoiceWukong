```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def compute_eer(fpr, tpr, thresholds):
    eer_index = np.nanargmin(np.abs(fpr - (1 - tpr)))
    eer = fpr[eer_index]
    eer_threshold = thresholds[eer_index]
    return eer, eer_threshold
```

```python
gat_zh_result='change this to your RawGAT-ST zh_eval_score.txt'
gat_en_result='change this to your RawGAT-ST en_eval_score.txt'
gat_column_name=['variant','audio_file','flag','source','label','score']
gat_zh_df=pd.read_csv(gat_zh_result,delimiter=' ',names=gat_column_name)
gat_en_df=pd.read_csv(gat_en_result,delimiter=' ',names=gat_column_name)
gat_zh_labels=gat_zh_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
gat_zh_scores=gat_zh_df['score']
gat_en_labels=gat_en_df['label'].apply(lambda x: 1 if x == 'bonafide' else 0)
gat_en_scores=gat_en_df['score']
gat_fpr_zh, gat_tpr_zh, gat_thresholds_zh = roc_curve(gat_zh_labels, gat_zh_scores)
gat_fpr_en, gat_tpr_en, gat_thresholds_en = roc_curve(gat_en_labels, gat_en_scores)
gat_eer_zh, gat_eer_threshold_zh = compute_eer(gat_fpr_zh, gat_tpr_zh, gat_thresholds_zh)
gat_eer_en, gat_eer_threshold_en = compute_eer(gat_fpr_en, gat_tpr_en, gat_thresholds_en)
gat_auc_zh = auc(gat_fpr_zh, gat_tpr_zh)
gat_auc_en = auc(gat_fpr_en, gat_tpr_en)
print(f'RawGAT-ST zh EER: {gat_eer_zh}, AUC: {gat_auc_zh}')
print(f'RawGAT-ST en EER: {gat_eer_en}, AUC: {gat_auc_en}')
```
